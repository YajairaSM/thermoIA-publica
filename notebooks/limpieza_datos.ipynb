## 1. Configuración del entorno

```python
import pandas as pd
import numpy as np

from pathlib import Path

# ============================================
# RUTAS
# ============================================

BASE_DIR = Path().resolve().parent

DATA_PROCESSED = BASE_DIR / "data" / "processed"

print("Ruta processed:")
print(DATA_PROCESSED)
```

## 2. Carga del dataset unido

```python
archivo = DATA_PROCESSED / "dataset_unido_2.csv"

df = pd.read_csv(archivo)

print("Dataset cargado correctamente")
print(f"Filas    : {len(df):,}")
print(f"Columnas : {len(df.columns)}")

print("\nColumnas:")
print(df.columns.tolist())
```

## 3. Inspección inicial

```python
# PRIMERAS FILAS
df.info()
```

```python
# Estadisticas descriptivas generales
df.describe()
```

```python
# ============================================
# CONVERTIR FECHAS
# ============================================

df["time"] = pd.to_datetime(df["time"])

print("Conversión realizada correctamente")
print(df["time"].dtype)
```

```python
# ============================================
# RANGO TEMPORAL
# ============================================

print("Fecha mínima:")
print(df["time"].min())

print("\nFecha máxima:")
print(df["time"].max())
```

```python
# Revision de nulos por columna antes de limpiar
nulos = df.isnull().sum()

print("Valores nulos por columna:")
print(nulos)
```

```python
# ============================================
# DUPLICADOS
# ============================================

duplicados = df.duplicated().sum()

print(f"Duplicados encontrados: {duplicados}")
```

```python
# ============================================
# ORDENAR CRONOLÓGICAMENTE
# ============================================

df = df.sort_values("time")

df = df.reset_index(drop=True)

print("Dataset ordenado correctamente")
```

## 4. Validación climática

```python
# ============================================
# VALIDACIÓN CLIMÁTICA
# ============================================

# Humedad inválida
humedad_invalida = df[
    (df["humidity"] < 0) |
    (df["humidity"] > 100)
]

# Lluvia negativa
lluvia_negativa = df[
    df["rain"] < 0
]

# Presión atmosférica inválida
presion_invalida = df[
    (df["pressure"] < 800) |
    (df["pressure"] > 1100)
]

# Temperaturas extremas
temperatura_extrema = df[
    (df["temperature"] < -10) |
    (df["temperature"] > 50)
]

print("VALIDACIÓN CLIMÁTICA")
print("-" * 40)

print(f"Humedad inválida      : {len(humedad_invalida)}")
print(f"Lluvia negativa       : {len(lluvia_negativa)}")
print(f"Presión inválida      : {len(presion_invalida)}")
print(f"Temperaturas extremas : {len(temperatura_extrema)}")
```

## 5. Variables temporales

```python
# ============================================
# VARIABLES TEMPORALES
# ============================================

df["year"] = df["time"].dt.year
df["month"] = df["time"].dt.month
df["day"] = df["time"].dt.day
df["hour"] = df["time"].dt.hour

print("Variables temporales creadas")
```

## 6. Registros por día

```python
# ============================================
# REGISTROS POR DÍA
# ============================================

df["date"] = df["time"].dt.date

registros_por_dia = df.groupby("date").size()

print(registros_por_dia.value_counts().head())
```

## 9. Guardar dataset limpio

```python
# ============================================
# GUARDAR DATASET LIMPIO
# ============================================

archivo_salida = DATA_PROCESSED / "dataset_limpio.csv"

df.to_csv(archivo_salida, index=False)

print(f"Dataset limpio guardado: {archivo_salida}")

print(f"\nFilas:")
print(f"{len(df):,}")

print(f"\nColumnas:")
print(df.columns.tolist())
```
