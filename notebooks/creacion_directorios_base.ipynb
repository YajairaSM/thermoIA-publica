# LIBRERIAS PARA CREAR DIRECTORIOS
pathlib: interfaz moderna y orientada a objetos para manejar rutas;
os: librería estándar para interacción con el sistema operativo.

```python
from pathlib import Path
import os
import shutil

# Verificar ubicacion actual
print("Ubicación del notebook :", Path.cwd())
print("Usuario                :", Path.home())

# Ruta raíz del proyecto
BASE_DIR = Path(r"C:\Users\Yajaira\Desktop\PROYECTO_PREDICCION_IA")

print("Ruta base del proyecto:")
print(BASE_DIR)
```

```python
# Definicion de todas las carpetas del proyecto
carpetas = [
    BASE_DIR / "data"      / "raw",         # CSV original de Open-Meteo
    BASE_DIR / "data"      / "processed",   # datos limpios y features
    BASE_DIR / "notebooks",                  # notebooks en orden numerado
    BASE_DIR / "src",                        # config.py, train.py, predict.py
    BASE_DIR / "models",                     # modelos .pkl entrenados
    BASE_DIR / "app",                        # app Streamlit
    BASE_DIR / "reports",                    # graficas y metricas
    BASE_DIR / "papers_latex",               # reporte o articulo final
]

# Crear cada carpeta
print("Creando estructura de carpetas...")
print()

for carpeta in carpetas:
    carpeta.mkdir(parents=True, exist_ok=True)
    nivel    = len(carpeta.relative_to(BASE_DIR).parts)
    sangria  = "    " * nivel
    print(f"{sangria}✓ {carpeta.relative_to(BASE_DIR)}")

print()
print("Estructura creada correctamente en:")
print(BASE_DIR)
```
